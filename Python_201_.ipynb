#Personal Expense tracker

> A simple Python program to track daily expenses.
It allows users to add, view, and calculate expenses by category and month along with search and delete options for the expenses.
The program also saves expenses to a text file and loads them when needed.

**How to run:** Run all cells in order, then call main() to start the expense tracker.

##1. Data Setup




*Categories*

```python
categories = [
    "Food",
    "Groceries",
    "Travel",
    "Rent",
    "Electricity",
    "Internet",
    "Repairs",
    "Health",
    "Shopping",
    "Education",
    "Entertainment",
    "Savings",
    "Loan EMI",
    "Other"
]

expenses = []
```

*Expenses*

```python
expenses = {
    "January": [],
    "February": [],
    "March": [],
    "April": [],
    "May": [],
    "June": [],
    "July": [],
    "August": [],
    "September": [],
    "October": [],
    "November": [],
    "December": []
}

monthly_budgets = {}
```

*Month Selection*

```python
def select_month():

  print("Select month:")
  months()

  month_choice = int(input("Enter month number: "))

 #-------Validate the month choice---------
  if month_choice < 1 or month_choice > 12:
    print("Invalid month number")
    return None

  return list(expenses.keys())[month_choice - 1]
```

##2. Add New Expense

```python
def months():
  for number,month in enumerate(expenses, start = 1):
        print(number,month)
def show_categories():
  for number,category in enumerate(categories,start = 1):
           print(number,category)
```

```python
def add_expense():

  # -------------Select month-----------------
  month = select_month()
  if month is None:
    return

  #--------------Select category--------------

  print("Select Category:")
  show_categories()

  category_choice = int(input("Enter category number: "))
  category = categories[category_choice - 1]

  #-----------Enter expense details------------

  amount = float(input("Enter amount: "))

  description = input("Enter description: ")

  date = input("Enter date (dd-mm-yy):")


  #-----------Create expense -----------------

  expense = {
    "category" : category,
    "amount"   : amount,
    "description" : description,
    "date" : date
}

  #----------------Add expense----------------

  expenses[month].append(expense)

  #-------------Confirmation message-----------

  print("Expense added successfully!")
```

##2. View Expenses

```python
def view_expenses():
# -------------Select month-----------------
  month = select_month()
  if month is None:
    return
     #Check whether the selected month has expenses
  if len(expenses[month]) == 0:
       print("No expenses found for this month")
       #Display all expenses for selected months
  else:
    for expense in expenses[month]:
       print("Date:",expense["date"])
       print("Category:",expense["category"])
       print("Amount: Rs",expense["amount"])
       print("Description:",expense["description"])
       print()


```

##3. Category-wise Totals (yearly and monthly)

```python
def category_totals():

    print("1. Monthly category totals")
    print("2. Yearly category totals")

    choice = int(input("Enter your choice: "))
    if choice < 1 or choice > 2:
      print("Invalid choice")
      return

#------------ monthly category totals-------------

    if choice == 1:
        # -------------Select month-----------------
           month = select_month()
           if month is None:
             return

           if len(expenses[month]) == 0:
            print("No expenses found for this month")
           else:
            for category in categories:
             total = 0
             for expense in expenses[month]:
               if expense["category"] == category:
                total = total + expense["amount"]
             print("Category:",category)
             print("Total: Rs",total)


#------------ Yearly category totals-------------------

    elif choice == 2:

     for category in categories:
        total = 0

        for month in expenses:
            for expense in expenses[month]:
                if expense["category"] == category:
                    total = total + expense["amount"]

        print("Category:",category)
        print("Total:Rs",total)
    else:
          print("Invalid choice")
```

##4. View Total Amount spent


```python
def total_amount_spent():

  total = 0
  for month in expenses:
    for expense in expenses[month]:
      total = total + expense["amount"]
  print("Total amount spent: Rs", total)
```

##5. File Handling


```python
def save_expenses():

  #open the file in write mode
  file = open("expenses.txt", "w")

  #Go through each month
  for month in expenses:
    file.write(month + "\n")                            #month name to the file
    for expense in expenses[month]:                     #expenses in the current month
      file.write("Date:" + expense["date"] + "\n")
      file.write("Category:" + expense["category"]+"\n")  #write expense details
      file.write ("Amount:"+ str(expense["amount"]) + "\n")
      file.write("Description"+ expense["description"])

  file.close()                                           #close file

  print("Expenses saved successfully!")
```

```python
def load_expenses():

  #open the file in read mode
  file = open("expenses.txt","r")

  #Read all lines from the file
  lines = file.readlines()

  #Getting all the months from expenses dictionary
  months_list = list(expenses.keys())

  #Go thorugh each line in the file
  for line in lines:
    line = line.strip()

  #Read and process each line from the file to seperate each month expenses
    if line in months_list:
      current_month = line
    elif line.startswith ("Date:"):
      date = line.replace("Date:","").strip()
    elif line.startswith("Category:"):
      category = line.replace("Category:","").strip()
    elif line.startswith("Amount:",):
      amount= float(line.replace("Amount:","").strip())
    elif line.startswith ("Description:"):
      description = line.replace("Description:","").strip()

  #Create an expense dictionary and add it to the current month
      expense = {
        "date" : date,
        "category" : category,
        "amount" : amount,
        "description" : description
  }

      expenses[current_month].append(expense)

  #closing the file
  file.close()

  print("Expenses loads successfully!!")


```

##6. Monthly budget

```python
def monthly_budget():
# -------------Select month-----------------
  month = select_month()
  if month is None:
    return

#Allocating the intended budget for particular month
  budget = float(input("Enter monthly budget: "))
  monthly_budgets[month] = budget
  print("Monthly budget set successfully!!")

#Calculating the total spending of the month selected
  total_spent = 0
  for expense in expenses[month]:
    total_spent = total_spent + expense["amount"]

#Comparing the amount spent in a month with monthly budget
  if total_spent > budget:
     print("⚠️ Budget exceeded!")
     print("Budget: Rs", budget)
     print("Total spent: Rs", total_spent)
     print("Exceeded by: Rs", total_spent - budget)
  else:
     print("Budget: Rs", budget)
     print("Total spent: Rs", total_spent)
     print("Budget remaining: Rs", budget - total_spent)
```

##7. Delete Expense

```python
def delete_expense():
# -------------Select month-----------------
  month = select_month()
  if month is None:
    return
  print("Select expense to delete:")

    # Display expenses with numbers
 if len(expenses[month]) == 0:
      print("No expenses found for this month.")
 else:
      for i, expense in enumerate(expenses[month], start=1):
        print(i, expense["date"], expense["category"], expense["amount"], expense["description"])

     # Get the expense number to delete
      expense_choice = int(input("Enter expense number to delete: "))

    if expense_choice < 1 or expense_choice > len(expenses[month]):
        print("Invalid expense number.")
    else:
        del expenses[month][expense_choice - 1]
        print("Expense deleted successfully!")
```

##8. Search Expense

```python
def search_expense():
# -------------Select month-----------------
  month = select_month()
  if month is None:
    return

#-------------Get Search keyword ------------
  print("Search Expense")
  search_term = input("Enter category, date or description to search :").lower()


#------------Look for matching expenses-------------
  found = False
  for expense in expenses[month]:
    if (search_term in expense["category"].lower()
        or search_term in expense["date"].lower()
        or search_term in expense["description"].lower()):
      print(expense)
      found = True

#-------------Dsiplay message if no match found--------
  if not found:
    print("No expenses found")
```

```python
def main():

  #-------------Options of main menu-------------
  while True:
    print("===== EXPENSE TRACKER =====")
    print("1. Add expense")
    print("2. View expenses")
    print("3. Category totals")
    print("4. Total amount spent")
    print("5. Save expenses")
    print("6. Load expenses")
    print("7. Monthly Budget")
    print("8. Delete Expense")
    print("9. Search expense")
    print("10. Exit")

    #-------------Pick a choice from main menu-------------
    choice = input("Enter your choice: ")

    if choice.isdigit():
      choice = int(choice)
    else:
      print("Please enter a number.")
      continue

    if choice == 1:
      add_expense()
    elif choice == 2:
      view_expenses()
    elif choice == 3:
      category_totals()
    elif choice == 4:
      total_amount_spent()
    elif choice == 5:
      save_expenses()
    elif choice == 6:
      load_expenses()
    elif choice == 7:
      monthly_budget()
    elif choice == 8:
      delete_expense()
    elif choice == 9:
      search_expense()
    elif choice == 10:
      print("Thank you for using Expense Tracker!!")
      break
    else:
      print("Invalid choice")


main()
```
