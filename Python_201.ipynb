#Personal Expense tracker

> This program helps users record and manage their daily expenses. Users can view all expenses and see how much they spend in each category. The program saves the expense data to a file so it can be used again when the program is reopened.

**How to run:** Run all cells, then call main() to start the program.

##1. Data Setup




*Categories*

```python
categories = [
    "Food",
    "Groceries",
    "Travel",
    "Rent",
    "Electricity",
    "Internet",
    "Repairs",
    "Health",
    "Shopping",
    "Education",
    "Entertainment",
    "Savings",
    "Loan EMI",
    "Other"
]

expenses = []
```

*Expenses*

```python
expenses = {
    "January": [],
    "February": [],
    "March": [],
    "April": [],
    "May": [],
    "June": [],
    "July": [],
    "August": [],
    "September": [],
    "October": [],
    "November": [],
    "December": []
}
```

##2. Add New Expense

```python
def months():
  for number,month in enumerate(expenses, start = 1):
        print(number,month)
def show_categories():
  for number,category in enumerate(categories,start = 1):
           print(number,category)
```

```python
def add_expense():

  # -------------Select month-----------------

  print("Select month:")
  months()

  month_choice = int(input("Enter month number: "))
  month = list(expenses.keys())[month_choice - 1]

  #--------------Select category--------------

  print("Select Category:")
  show_categories()

  category_choice = int(input("Enter category number: "))
  category = categories[category_choice - 1]

  #-----------Enter expense details------------

  amount = float(input("Enter amount: "))

  description = input("Enter description: ")

  date = input("Enter date (dd-mm-yy):")


  #-----------Create expense -----------------

  expense = {
    "category" : category,
    "amount"   : amount,
    "description" : description,
    "date" : date
}

  #----------------Add expense----------------

  expenses[month].append(expense)

  #-------------Confirmation message-----------

  print("Expense added successfully!")
```

```python
add_expense()      #Enter the details of the expense
```

##2. View Expenses

```python
def view_expenses():
  # -------------Select month-----------------

  print("Select month:")
  months()

  month_choice = int(input("Enter month number: "))

  if month_choice < 1 or month_choice > 12 :
    print("Invalid month number")
  else:
     month = list(expenses.keys())[month_choice - 1]
     if len(expenses[month]) == 0:
       print("No expenses found for this month")
     else:
      for expense in expenses[month]:
       print("Date:",expense["date"])
       print("Category:",expense["category"])
       print("Amount: Rs",expense["amount"])
       print("Description:",expense["description"])
       print()


```

```python
view_expenses()
```
