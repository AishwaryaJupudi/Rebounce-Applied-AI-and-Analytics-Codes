#Part - 2

##Missing customer id

Missing CustomerID values were retained because CustomerID is not required for the sales and country-level analysis.

##Converting invoice date from text to a proper datetime column

```python
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
```

```python
df["InvoiceDate"].dtype    #dtype('<M8[ns]') is internal representation of datetime
```

##Removing the duplicate rows

```python
df = df.drop_duplicates()
```

```python
df.shape       #checking the shape after dropping  duplicates
```

```python
df.duplicated().sum()    #checking the duplicate count after cleaning
```

##Removing invalid transactions


```python
df = df[(df['Quantity']>0) & (df['UnitPrice']>0)]
```

```python
df.shape
```

##Creating a new Sales column using the formula
    Sales = Quantity × UnitPrice

```python
df['Sales'] = df['Quantity'] * df['UnitPrice']
```
