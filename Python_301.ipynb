#Part - 2 Clean the data

```python
df_clean = df.copy()        # Create a copy of the original dataset for cleaning and analysis
```

##Missing customer id

Missing CustomerID values were retained because dropping 135080 transactions could affect the analysis and also CustomerID is not required for the sales and country-level analysis.

##Converting invoice date from text to a proper datetime column

```python
df_clean['InvoiceDate'] = pd.to_datetime(df_clean['InvoiceDate'])
```

```python
df_clean["InvoiceDate"].dtype
```

##Removing the duplicate rows

```python
df_clean = df_clean.drop_duplicates()
```

```python
df_clean.shape       #checking the shape after dropping  duplicates
```

```python
df_clean.duplicated().sum()    #checking the duplicate count after cleaning
```

##Removing invalid transactions


```python
df_clean = df_clean[(df_clean['Quantity']>0) & (df_clean['UnitPrice']>0)]
```

```python
df_clean.shape
```

##Creating a new Sales column using the formula
    Sales = Quantity × UnitPrice

```python
df_clean['Sales'] = df_clean['Quantity'] * df_clean['UnitPrice']
```

```python
df_clean.head()
```

##Rounding the values to a workable number of decimal places

```python
print("Mean:", round(np.mean(df_clean['Sales']),2))
print("Median:",round(np.median(df_clean['Sales']),2))
print("Min:",round(np.min(df_clean['Sales']),3))
print("Max:",round(np.max(df_clean['Sales']),2))
print("Standard Deviation:",round(np.std(df_clean['Sales']),2))
```

The mean sales value (20.28) is considerably higher than the median (9.92), indicating that the sales distribution is right-skewed. This suggests that a small number of very large transactions are pulling the mean upward, while most transactions have relatively lower sales values.

#Part - 5  GroupBy and aggregation

##Building a summary table with each country in a row

```python
country_summary = df_clean.groupby('Country').agg(
    Number_of_transactions= ('Country', 'count'),
    Total_sales=('Sales','sum'),
    Average_sales=('Sales','mean')
).reset_index()
```

```python
#Rounding the decimal values of Total sales and average sales to 2 decimal places
country_summary['Total_sales']=country_summary['Total_sales'].round(2)
country_summary['Average_sales']=country_summary['Average_sales'].round(2)
```

```python
country_summary
```

##Five countries that generated the highest total sales

```python
top_five_countries = country_summary.sort_values('Total_sales',ascending=False).head()
top_five_countries
```

##Country that has the highest average transaction value

```python
country_with_highest_avg = country_summary.sort_values('Average_sales',ascending=False).head(1)
country_with_highest_avg
```

##Number of Transactions of the top-selling country

```python
Top_country = country_summary.sort_values('Total_sales',ascending=False).head(1)
Top_country
```

```python
Top_country['Number_of_transactions'].iloc[0]
```

1. The United Kingdom generated the highest total sales, mainly due to its very large number of transactions (479,985), despite having a lower average transaction value of 18.75.
2. The Netherlands had the highest average transaction value (121.00), showing that a high total sales value does not necessarily mean a high average transaction value.

Overall, the results demonstrate that a country's total sales are strongly influenced by transaction volume, whereas average sales reflect the typical value of an individual transaction. Therefore, a country can have a lower total sales value but a much higher average transaction value, as observed for the Netherlands.

#Part - 6 Merge and Region analysis

```python
region_lookup = {
"United Kingdom": "Europe",
"France": "Europe",
"Germany": "Europe",
"Spain": "Europe",
"Netherlands": "Europe",
"Belgium": "Europe",
"Switzerland": "Europe",
"Portugal": "Europe",
"Australia": "Oceania",
"Japan": "Asia",
"USA": "Americas",
}
```

```python
region_df = pd.DataFrame(region_lookup.items(),columns=['Country','Region'])
region_df
```

##Merge it onto country summary table using a left merge on Country.

```python
country_region = country_summary.merge(
    region_df,
    on='Country',
    how='left'
)
```

```python
country_region
```

##Filling the gaps in Region with "Other"

```python
country_region["Region"]=country_region["Region"].fillna("Other")
country_region
```

##Group by Region and find the total sales per region.

```python
region_sales = country_region.groupby('Region')['Total_sales'].sum()
region_sales
```

##Which region generated the highest total sales?

```python
region_sales.sort_values(ascending=False)
```

Europe generated the highest total sales (9918999.75), mainly because several high-sales countries, including the United Kingdom, are part of this region
