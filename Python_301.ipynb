#Part - 2 Clean the data

```python
df_clean = df.copy()        # Create a copy of the original dataset for cleaning and analysis
```

##Missing customer id

Missing CustomerID values were retained because dropping 135080 transactions could affect the analysis and also CustomerID is not required for the sales and country-level analysis.

##Converting invoice date from text to a proper datetime column

```python
df_clean['InvoiceDate'] = pd.to_datetime(df_clean['InvoiceDate'])
```

```python
df_clean["InvoiceDate"].dtype    #dtype('<M8[ns]') is internal representation of datetime
```

##Removing the duplicate rows

```python
df_clean = df_clean.drop_duplicates()
```

```python
df_clean.shape       #checking the shape after dropping  duplicates
```

```python
df_clean.duplicated().sum()    #checking the duplicate count after cleaning
```

##Removing invalid transactions


```python
df_clean = df_clean[(df_clean['Quantity']>0) & (df['UnitPrice']>0)]
```

```python
df_clean.shape
```

##Creating a new Sales column using the formula
    Sales = Quantity × UnitPrice

```python
df_clean['Sales'] = df_clean['Quantity'] * df_clean['UnitPrice']
```

```python
df_clean.head()
```

#Part - 3 Numpy Analysis

##Calculating the mean

##Calculating the median

##Calculating the minimum sales

```python
np.min(df_clean['Sales'])
```

##Calculating the maximum sales

```python
np.max(df_clean['Sales'])
```

#Calculating the standard deviation

```python
np.std(df_clean['Sales'])
```

##Rounding the values to a workable number of decimel places

```python
print("Mean:", round(np.mean(df_clean['Sales']),2))
print("Median:",round(np.median(df_clean['Sales']),2))
print("Min:",round(np.min(df_clean['Sales']),3))
print("Max:",round(np.max(df_clean['Sales']),2))
print("Standard Deviation:",round(np.std(df_clean['Sales']),2))
```

The mean sales value (20.28) is considerably higher than the median (9.92), indicating that the sales distribution is right-skewed. This suggests that a small number of very large transactions are pulling the mean upward, while most transactions have relatively lower sales values.
