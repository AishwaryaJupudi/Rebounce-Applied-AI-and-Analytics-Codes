Problem-5 ***The password strength check***

```python
password     = "AppliedAI@1234"
length       = len(password)
length_greater_than_8 = length >= 8
contains_at       = "@" in password
is_it_strong = length >= 8 and contains_at
print("Password:", password)
print("Length is OK(>=8):", length_greater_than_8)
print("Contains @:", contains_at)
print("Password is strong:", is_it_strong)
```

Problem-6 ***The Class Roster***

---



```python
attendees = ["Aishwarya","Vijay","Varun","Anuja","Vishnu","Swetha","Monika","Sreya"]
total_attendees = len(attendees)
first_attendee  = attendees[0]
last_two_attendees = attendees[-2:]
first_3_attendees_alphabetically = sorted(attendees)[:3]
print("Total attendees:",total_attendees)
print("First attendee:",first_attendee)
print("Last two attendees:",last_two_attendees)
print("First three attendees alphabetically",first_3_attendees_alphabetically)
```
